### Part 1. 리스트와 반복문을 이용한 데이터 처리

여러 개의 데이터를 리스트에 저장하고 `for` 반복문을 이용하여 하나씩 처리합니다.

다음과 같은 작업을 수행합니다.

- 리스트 생성
- 리스트의 데이터 순회
- 조건에 맞는 데이터 찾기
- 조건을 만족하는 데이터 개수 계산
- 계산 결과 출력

**필수 역량**

`list`, `for`, `if`

````python
scores = [65, 70, 90, 10, 30]

count = 0
for score in scores:
    if score >= 70:
        count += 1
        
print(f"70점 이상 학생 수: {count}명")
````

### art 2. 딕셔너리를 이용한 단어 빈도 분석

다음과 같은 문장을 대상으로 단어가 몇 번 등장하는지 분석합니다.

```
sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""
```

다음 과정을 직접 구현합니다.

1. `split()`으로 문장을 단어 단위로 나눕니다.
2. 빈 딕셔너리를 만듭니다.
3. 단어를 하나씩 반복합니다.
4. 처음 등장한 단어는 딕셔너리에 `1`로 저장합니다.
5. 이미 존재하는 단어는 기존 값에 `1`을 증가시킵니다.
6. 전체 단어와 등장 횟수를 출력합니다.
7. 일정 횟수 이상 등장한 단어만 별도로 출력합니다.

예:

```
파이썬 : 4회
공부는 : 2회
...
```

**필수 조건**

- `for` 사용
- `if` 사용
- `dict` 사용
- `collections.Counter` 등 자동 빈도 계산 기능은 사용하지 않음

**필수 역량**

`str`, `split()`, `dict`, `for`, `if`

````python
sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""

words = sentence.split()

word_count = {}
for word in words:
    if word not in word_count:
        word_count[word] = 1
    else:
        word_count[word] += 1
print(word_count)

for word in word_count:
    if word_count[word] >= 2:
        print(f"{word}: {word_count[word]}회")
````

### Part 3. 함수로 반복되는 코드 정리하기

Part 2에서 작성한 단어 빈도 분석 기능을 함수로 만들어봅니다.

예:

```
defcount_words(sentence):# 단어 빈도를 계산하는 코드
    ...
```

다음 내용을 확인합니다.

- 함수에 문자열 전달
- 함수 내부에서 단어 빈도 계산
- 계산 결과 반환
- 반환된 결과 출력

가능하다면 특정 횟수 이상 등장한 단어만 출력하는 기능도 별도 함수로 분리해봅니다.

**필수 역량**

`def`, 매개변수, `return`

````python
def count_word(sentence):
    words = sentence.split()

    word_count = {}
    for word in words:
        if word not in word_count:
            word_count[word] = 1
        else:
            word_count[word] += 1
    return word_count

def many_count(word):
    for word in word_count:
        if word_count[word] >= 2:
            print(f"{word}: {word_count[word]}회")
sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""
result = count_word(sentence)
print(result)
many_count(result)
````
